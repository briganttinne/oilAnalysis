```python
import pandas as pd

# Load the pipeF.csv file into a DataFrame
df = pd.read_csv('pipeF.csv')

# Strip any whitespace from column names
df.columns = df.columns.str.strip()

# Convert 'us', 'Alaska', and 'GOM' columns to numeric, handling commas and potential non-numeric values
df['us'] = pd.to_numeric(df['us'].str.replace(',', ''), errors='coerce')
df['Alaska'] = pd.to_numeric(df['Alaska'], errors='coerce')
df['GOM'] = pd.to_numeric(df['gom'].str.replace(',', ''), errors='coerce') # Assuming 'gom' is the correct column name based on kernel state

# Calculate 'L48 volume' by subtracting 'Alaska' and 'GOM' from 'us'
df['L48 volume'] = df['us'] - df['Alaska'] - df['GOM']
```

### Loading Data from EIA Excel File

Now, I will load the data from the provided URL into a new pandas DataFrame called `eia_df`. I'm assuming the data starts on 'Data 1' sheet and skips the first 2 rows, which is a common format for EIA's Excel files. If this doesn't look right, please let me know!

```python
eia_url = 'https://www.eia.gov/dnav/pet/xls/PET_CRD_CRPDN_ADC_MBBL_M.xls'

try:
    # Attempt to read the Excel file from the URL
    # Common sheet name is 'Data 1' and data usually starts after a few header rows (e.g., skip 2 rows)
    eia_df = pd.read_excel(eia_url, sheet_name='Data 1', skiprows=2)
    display(eia_df.head())
except Exception as e:
    print(f"Error loading data: {e}")
    print("Please check the URL, sheet name, or skiprows argument if the data doesn't load correctly.")
```

### Filtering Data Before 2010

Now, I'll convert the 'Date' column to datetime objects and then filter out all rows with dates before January 1, 2010.

```python
# Convert the 'Date' column to datetime objects
# The column name 'Date' from the eia_df.head() output seems correct.
# Using errors='coerce' to turn any unparseable dates into NaT (Not a Time).
eia_df['Date'] = pd.to_datetime(eia_df['Date'], errors='coerce')

# Filter out rows where 'Date' is before 2010
eia_df_filtered = eia_df[eia_df['Date'].dt.year >= 2010].copy()

# Display the head of the filtered DataFrame to confirm the change
display(eia_df_filtered.head())
```

### Converting Dates to Month-End

I will now adjust the 'Date' column in `eia_df_filtered` to represent the last day of each month. This is useful for consistent monthly aggregations or reporting.

```python
# Convert the 'Date' column to the last day of the month
eia_df_filtered['Date'] = eia_df_filtered['Date'].dt.to_period('M').dt.end_time

# Display the head of the modified DataFrame to confirm the change
display(eia_df_filtered.head())
```

### Dropping Time Component from Date Column

I will now convert the 'Date' column from datetime objects to just date objects, removing the time component.

```python
# Drop the time component from the 'Date' column
eia_df_filtered['Date'] = eia_df_filtered['Date'].dt.date

# Display the head of the modified DataFrame to confirm the change
display(eia_df_filtered.head())
```

```python
eia_df_filtered.info()
```

```python
# Select only the specified columns by their names
# Column 0: 'Date'
# Column 1: 'U.S. Field Production of Crude Oil (Thousand Barrels)'
# Column 28: 'Texas Field Production of Crude Oil (Thousand Barrels)'
# Column 36: 'Alaska Field Production of Crude Oil (Thousand Barrels)'
eia_df_selected = eia_df_filtered[[
    'Date',
    'U.S. Field Production of Crude Oil (Thousand Barrels)',
    'Federal Offshore--Gulf of America Field Production of Crude Oil (Thousand Barrels)',
    'Alaska Field Production of Crude Oil (Thousand Barrels)'
]].copy()

display(eia_df_selected.head())
```

```python
# Rename the columns as requested
eia_df_selected = eia_df_selected.rename(columns={
    'U.S. Field Production of Crude Oil (Thousand Barrels)': 'US',
    'Federal Offshore--Gulf of America Field Production of Crude Oil (Thousand Barrels)': 'GOA',
    'Alaska Field Production of Crude Oil (Thousand Barrels)': 'AK'
})

# Display the head of the modified DataFrame to confirm the changes
display(eia_df_selected.head())
```

```python
# Set the 'Date' column as the index
eia_df_selected = eia_df_selected.set_index('Date')

# Display the head of the DataFrame to confirm the change
display(eia_df_selected.head())
```

```python
eia_df_selected['L48'] = eia_df_selected['US'] - eia_df_selected['GOA'] - eia_df_selected['AK']

display(eia_df_selected.head())
```

```python
# Calculate 'days_in_month' directly from the index of eia_df_selected.
# The index currently contains datetime.date objects, so convert them to datetime.datetime first.
eia_df_selected['days_in_month'] = pd.to_datetime(eia_df_selected.index).daysinmonth

# Convert US, GOA, AK, and L48 to daily rates
eia_df_selected['US (per Day)'] = eia_df_selected['US'] / eia_df_selected['days_in_month']
eia_df_selected['GOA (per Day)'] = eia_df_selected['GOA'] / eia_df_selected['days_in_month']
eia_df_selected['AK (per Day)'] = eia_df_selected['AK'] / eia_df_selected['days_in_month']
eia_df_selected['L48 (per Day)'] = eia_df_selected['L48'] / eia_df_selected['days_in_month']

# Drop the original monthly columns and the 'days_in_month' column
eia_df_selected = eia_df_selected.drop(columns=['US', 'GOA', 'AK', 'L48', 'days_in_month'])

# Display the head of the DataFrame to confirm the change
display(eia_df_selected.head())
```

```python
# Rename columns to reflect 'Mb/d' unit
eia_df_selected = eia_df_selected.rename(columns={
    'US (per Day)': 'US (Mb/d)',
    'GOA (per Day)': 'GOA (Mb/d)',
    'AK (per Day)': 'AK (Mb/d)',
    'L48 (per Day)': 'L48 (Mb/d)'
})

# Display the head of the DataFrame to confirm the change
display(eia_df_selected.head())
```

```python
# Display the first 5 rows with the new 'L48 volume' column
display(df.head())
```

```python
# Convert 'Date' column to datetime objects
df['Date'] = pd.to_datetime(df['Date'])

# Set 'Date' as the index for resampling
df_indexed = df.set_index('Date')

# Resample to monthly frequency and calculate the mean for numeric columns
df_monthly_avg = df_indexed[['us', 'Alaska', 'GOM', 'L48 volume']].resample('M').mean().reset_index()

# Display the first few rows of the monthly average data
display(df_monthly_avg.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(15, 8))

sns.lineplot(x='Date', y='us', data=df_monthly_avg, label='US Volume')
sns.lineplot(x='Date', y='Alaska', data=df_monthly_avg, label='Alaska Volume')
sns.lineplot(x='Date', y='GOM', data=df_monthly_avg, label='GOM Volume')
sns.lineplot(x='Date', y='L48 volume', data=df_monthly_avg, label='L48 Volume (US - Alaska - GOM)')

plt.title('Monthly Average Volumes Over Time')
plt.xlabel('Date')
plt.ylabel('Average Volume')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

### Merging Datasets

Now, I will merge the `df_monthly_avg` (from `pipeF.csv`) and `eia_df_selected` (from EIA) DataFrames. Both contain monthly aggregated data, with the 'Date' column representing the end of the month. I will rename the columns from `eia_df_selected` to clearly indicate their source (e.g., `US_EIA (Mb/d)`) and then perform an outer merge on the 'Date' column.

```python
# Convert the index of eia_df_selected to DatetimeIndex
eia_df_selected.index = pd.to_datetime(eia_df_selected.index)

# Reset index of eia_df_selected to make 'Date' a column for merging
# Now 'Date' column will automatically be datetime64[ns]
eia_df_selected_reset = eia_df_selected.reset_index()

# Rename columns of eia_df_selected_reset to distinguish them from df_monthly_avg's columns
eia_df_selected_reset = eia_df_selected_reset.rename(columns={
    'US (Mb/d)': 'US_EIA (Mb/d)',
    'GOA (Mb/d)': 'GOA_EIA (Mb/d)',
    'AK (Mb/d)': 'AK_EIA (Mb/d)',
    'L48 (Mb/d)': 'L48_EIA (Mb/d)'
})

# Merge the two monthly DataFrames on the 'Date' column using an outer merge
merged_df = pd.merge(df_monthly_avg, eia_df_selected_reset, on='Date', how='outer')

# Display the head of the merged DataFrame
display(merged_df.head())
```

### Calculate Difference between US_EIA and us

Now I will calculate the difference between the 'US_EIA (Mb/d)' and 'us' columns and store it in a new column named 'US_D'.

```python
# Calculate the difference between 'US_EIA (Mb/d)' and 'us' columns
merged_df['US_D'] = merged_df['US_EIA (Mb/d)'] - merged_df['us']
merged_df['AK_D'] = merged_df['AK_EIA (Mb/d)'] - merged_df['Alaska']
merged_df['GOA_D'] = merged_df['GOA_EIA (Mb/d)'] - merged_df['GOM']
merged_df['L48_D'] = merged_df['L48_EIA (Mb/d)'] - merged_df['L48 volume']

# Display the head of the DataFrame with the new 'US_D' column
display(merged_df.head())
```

### Plotting GOM, L48 Volume, GOA_EIA (Mb/d), and L48_EIA (Mb/d)

I will now create a plot to visualize the trends of 'GOM', 'L48 volume', 'GOA_EIA (Mb/d)', and 'L48_EIA (Mb/d)' from the merged dataset. This will help us compare the different sources and regions over time.

```python
import plotly.express as px

# Create a line plot for the specified columns
fig = px.line(
    merged_df,
    x='Date',
    y=['GOM', 'L48 volume', 'GOA_EIA (Mb/d)', 'L48_EIA (Mb/d)'],
    labels={
        'value': 'Production Volume (Mb/d)',
        'variable': 'Region/Source'
    },
    title='Comparison of GOM, L48 Volume, GOA_EIA, and L48_EIA Over Time'
)

# Update legend names for clarity
fig.for_each_trace(lambda t: t.update(name = t.name
                                       .replace('GOM', 'GOM (PipeF)')
                                       .replace('L48 volume', 'L48 Volume (PipeF)')
                                       .replace('GOA_EIA (Mb/d)', 'GOA (EIA)')
                                       .replace('L48_EIA (Mb/d)', 'L48 (EIA)')))

fig.show()
```
